```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import warnings
warnings.filterwarnings("ignore")
```

```python


df = pd.read_csv("/content/hotel_bookings.csv")

print("Dataset Loaded Successfully!")

df.head()
```

```python
# Shape

print("Shape of Dataset :", df.shape)

# Column Names

print("\nColumns\n")
print(df.columns)

# Data Types

print("\nData Types\n")
print(df.dtypes)

# Missing Values

print("\nMissing Values\n")
print(df.isnull().sum())

# Duplicate Rows

print("\nDuplicate Rows :", df.duplicated().sum())
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
print("Duplicate Rows Before :", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate Rows After :", df.duplicated().sum())
```

```python
# Display Missing Values

missing = df.isnull().sum()

missing = missing[missing > 0]

print(missing)
```

```python
# Fill children with median

if 'children' in df.columns:
    df['children'].fillna(df['children'].median(), inplace=True)

# Fill country with mode

if 'country' in df.columns:
    df['country'].fillna(df['country'].mode()[0], inplace=True)

# Fill agent with 0

if 'agent' in df.columns:
    df['agent'].fillna(0, inplace=True)

# Fill company with 0

if 'company' in df.columns:
    df['company'].fillna(0, inplace=True)

print(df.isnull().sum())
```

```python
# Total Guests

df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

# Total Stay

df["total_stay"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

# Total Cost

df["total_cost"] = (
    df["adr"] *
    df["total_stay"]
)

df.head()
```

```python
# Label Encoding

encoder = LabelEncoder()

cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = encoder.fit_transform(df[col].astype(str))

print("Encoding Completed")
```

```python
X = df.drop("is_canceled", axis=1)

y = df["is_canceled"]

print(X.shape)

print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train.shape)

print(X_test.shape)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
# ================================
# Import Machine Learning Models
# ================================

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier,
    BaggingClassifier
)

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
```

```python
# ================================
# Machine Learning Models
# ================================

models = {

    "Logistic Regression": LogisticRegression(max_iter=1000),

    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "Random Forest": RandomForestClassifier(random_state=42),

    "Extra Trees": ExtraTreesClassifier(random_state=42),

    "KNN": KNeighborsClassifier(),

    "SVM": SVC(probability=True),

    "Naive Bayes": GaussianNB(),

    "MLP": MLPClassifier(max_iter=500),

    "Bagging": BaggingClassifier(random_state=42),

    "AdaBoost": AdaBoostClassifier(random_state=42),

    "Gradient Boosting": GradientBoostingClassifier(random_state=42)

}
```

```python
# ================================
# Train Every Model
# ================================

results = []

for name, model in models.items():

    print("="*60)
    print(name)

    # Train
    model.fit(X_train, y_train)

    # Predictions
    y_pred = model.predict(X_test)

    # Probabilities
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:,1]
    else:
        y_prob = None

    # Accuracy
    train_acc = model.score(X_train, y_train)

    test_acc = model.score(X_test, y_test)

    precision = precision_score(y_test, y_pred)

    recall = recall_score(y_test, y_pred)

    f1 = f1_score(y_test, y_pred)

    accuracy = accuracy_score(y_test, y_pred)

    if y_prob is not None:
        roc = roc_auc_score(y_test, y_prob)
    else:
        roc = np.nan

    # Cross Validation
    cv = cross_val_score(model, X_train, y_train, cv=5).mean()

    # Overfitting Check
    diff = train_acc - test_acc

    if diff <= 0.05:
        status = "Good Fit"

    elif diff <= 0.10:
        status = "Slight Overfit"

    else:
        status = "Overfitting"

    results.append([

        name,
        train_acc,
        test_acc,
        accuracy,
        precision,
        recall,
        f1,
        roc,
        cv,
        diff,
        status

    ])

    print("Train Accuracy :", round(train_acc,4))
    print("Test Accuracy  :", round(test_acc,4))
    print("Precision      :", round(precision,4))
    print("Recall         :", round(recall,4))
    print("F1 Score       :", round(f1,4))
    print("ROC AUC        :", round(roc,4) if not np.isnan(roc) else "N/A")
    print("CV Score       :", round(cv,4))
    print("Status         :", status)
```

```python
results_df = pd.DataFrame(results, columns=[

    "Model",
    "Train Accuracy",
    "Test Accuracy",
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC AUC",
    "Cross Validation",
    "Difference",
    "Status"

])

results_df
```

```python
results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

results_df.reset_index(drop=True, inplace=True)

results_df
```

```python
results_df.to_csv("ML_Model_Comparison.csv", index=False)

print("Results Saved Successfully")
```

```python
# ==========================================
# Feature Selection using SelectKBest
# ==========================================

from sklearn.feature_selection import SelectKBest, f_classif

# Select Top 15 Features
selector = SelectKBest(score_func=f_classif, k=15)

X_new = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]

print("Selected Features:\n")
print(selected_features)
```

```python
# ==========================================
# Train Test Split
# ==========================================

X_train_fs, X_test_fs, y_train_fs, y_test_fs = train_test_split(
    X_new,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Scaling
scaler_fs = StandardScaler()

X_train_fs = scaler_fs.fit_transform(X_train_fs)

X_test_fs = scaler_fs.transform(X_test_fs)
```

```python
# ==========================================
# Train Models on Selected Features
# ==========================================

feature_results = []

for name, model in models.items():

    print("="*60)
    print(name)

    model.fit(X_train_fs, y_train_fs)

    pred = model.predict(X_test_fs)

    if hasattr(model, "predict_proba"):
        prob = model.predict_proba(X_test_fs)[:,1]
        roc = roc_auc_score(y_test_fs, prob)
    else:
        roc = np.nan

    train_acc = model.score(X_train_fs, y_train_fs)
    test_acc = model.score(X_test_fs, y_test_fs)

    acc = accuracy_score(y_test_fs, pred)

    precision = precision_score(y_test_fs, pred)

    recall = recall_score(y_test_fs, pred)

    f1 = f1_score(y_test_fs, pred)

    cv = cross_val_score(model, X_train_fs, y_train_fs, cv=5).mean()

    diff = train_acc - test_acc

    if diff <= 0.05:
        status = "Good Fit"

    elif diff <= 0.10:
        status = "Slight Overfit"

    else:
        status = "Overfitting"

    feature_results.append([
        name,
        train_acc,
        test_acc,
        acc,
        precision,
        recall,
        f1,
        roc,
        cv,
        diff,
        status
    ])

    print("Accuracy :", round(acc,4))
```

```python
feature_results_df = pd.DataFrame(

    feature_results,

    columns=[

        "Model",

        "Train Accuracy",

        "Test Accuracy",

        "Accuracy",

        "Precision",

        "Recall",

        "F1",

        "ROC",

        "Cross Validation",

        "Difference",

        "Status"

    ]

)

feature_results_df = feature_results_df.sort_values(
    by="Accuracy",
    ascending=False
)

feature_results_df.reset_index(drop=True, inplace=True)

feature_results_df
```
